single inheritance

```python
class univ():
    def univ_name(self):
        print("jain university")

class student(univ):
    def student_name(Self):
        print("aashin")

s = student()

s.univ_name()
s.student_name()
```

multiple inheritance

```python
class university():
    def univ_name(Self):
        print("Jain UNiversity")

class dep():
    def department(self):
        print("MCA")

class student_detail(university, dep):
    def detail(self):
        print("Aashin")

std = student_detail()

std.univ_name()
std.department()
std.detail()
```

multilevel inheritance

```python
class uni():
    def university(self):
        print("Jain University")

class depart(uni):
    def department_name(self):
        print("MCA")

class std_name(depart):
    def name_dtls(Self):
        print("aashinmn")

s1 = std_name()

s1.university()
s1.department_name()
s1.name_dtls()
```

hierarchical inheritance


```python
class unive():
    def u_name(self):
        print("Jain University Bangalore")

class std_name1(unive):
    def s_name1(self):
        print("aashin")

class std_name2(unive):
    def s_name2(self):
        print("Kavin")

n1 = std_name1()
n2 = std_name2()

n1.u_name()
n1.s_name1()

n2.u_name()
n2.s_name2()
```

Hybrid inheritance

```python
class university():
    def uni_name(self):
        print("Jain University")

class department(university):
    def dep(self):
        print("MCA")

class year(university):
    def yr(self):
        print("2023 - 2025")

class name(department, year):
    def student_name(self):
        print("Aashin")

n1 = name()

n1.uni_name()
n1.dep()
n1.yr()
n1.student_name()

# mixed of two ir more inheritance is called hybrid inheritance
```
