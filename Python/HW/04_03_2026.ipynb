```python
def calculator():
    try:
        num1 = float(input("Enter a num: "))
        op = input("Choose +, -, *, / : ")
        num2 = float(input("Enter a num: "))

        if op == "+":
            print(num1 + num2)
        elif op == "-":
            print(num1 - num2)
        elif op == "*":
            print(num1 * num2)
        elif op == "/":
            # if num2 != 0:
                print(num1 / num2)
            # else:
            #     print("this is not divisible by zero")
        else:
            print("invalid operator")
    except Exception as e:
        print(e)
calculator()
```
