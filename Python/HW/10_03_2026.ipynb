```python
class person():
    def __init__(self, name, age = 22):# here the age is default as 22
        self.name = name
        self.age = age

per = person("aashin", 23)# age is edited as 23 here
per1 = person("kavin")

print(per.name)
print(per.age)
print(per1.name)
print(per1.age)
```

```python
class brand():
    def __init__(self, model, year):
        self.model = model
        self.year = year

    def greet(self):
        print("my fav car is : " + self.model)

b1 = brand("bmw_m5_competition", 2020)
b1.greet()
print(b1.year)
```
