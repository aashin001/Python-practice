```python
num = int(input("enter the num: "))
rev = 0

while num> 0:
    ans = num % 10
    rev = rev * 10 + ans
    num = num // 10

print(rev)
```

```python
pall = input("enter string")

rev = ""

for i in pall:
    rev = i + rev

if rev == pall:
    print("its a pallindrom")
else:
    print("its not pallindrom")
```

```python
la = list(map(int, input().split()))

largest = la[0]

for i in la:
    if i > largest:
        largest = i

print(largest)
```

```python
num = int(input("enter a num : "))
fact = 1

for i in range(1, num+1):
    fact = fact * i

print(fact)
```

```python
pri = int(input("enter a num"))

for i in range(2, pri):
    if pri % i == 0:
        print("its not a prime")
        break
else:
     print("its a prime")
    
```

```python
number = int(input("num"))
temp = number
sum = 0

while temp> 0:
    digit = temp % 10
    sum = sum + digit ** 3
    temp = temp // 10


if sum == number:
    print("armstrong")
else:
    print("not armstrong")
```
