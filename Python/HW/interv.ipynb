```python
a,b = map(int, input().split())

while b != 0:
    temp = b
    b = a % b
    a = temp

print(a)
```

```python
st = input("inp")

s = []
for i in st:
    s.append(i)

rev = ""

while s:
    rev += s.pop()

print(rev)
```
