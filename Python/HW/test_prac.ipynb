```python
marks = [85,40,77,92,60]

grade = []

for i in marks:
    if i >= 80:
        grade.append("A")
    elif i >= 70:
        grade.append("B")
    elif i >= 60:
        grade.append("C")
    elif i >= 50:
        grade.append("D")
    else:
        grade.append("F")
print(grade)
```
