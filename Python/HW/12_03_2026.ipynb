```python
inp1 = input("Enter the first word").lower()
inp2 = input("Enter the second word").lower()

def anagram(a, b):
    if len(a) != len(b):
        return "its not a anagram"

    count={}

    for i in range(len(a)):
        count[a[i]] = count.get(a[i],0) +1
        count[b[i]] = count.get(b[i],0) -1

    if all(val == 0 for val in count.values()):
        return "anagram"
    else:
        return "not anagram"




print(anagram(inp1, inp2))
```
