Del Update Add using dict

```python
car_brands = { "BMW" : "F90", "Skoda" : "VRS", "VW" : "Passat" }
```

Add

```python
car_brands["Audi"] = "RS7"
print(car_brands)
```

Update

```python
car_brands["Skoda"] = "Octovia VRS"
car_brands
```

Delete

```python
del car_brands["VW"]
car_brands
```
